# Guia de dicionários em Python 3 e Django 6
Um dicionário (dict) é uma estrutura que guarda pares chave → valor, como uma agenda telefônica: você procura pelo nome (chave) e recebe o número (valor). Em Python, ele é mutável, permite acesso rápido por chave e preserva a ordem em que os itens foram inseridos.


### 1 A ideia básica

```python
# Um dicionário é escrito entre chaves: {chave: valor}
pessoa = {
    "nome": "Ana",          # chave: "nome" | valor: "Ana"
    "idade": 30,            # chave: "idade" | valor: 30
    "cidade": "Recife",     # chave: "cidade" | valor: "Recife"
}

# Acessamos um valor usando a chave entre colchetes
print(pessoa["nome"])       # Saída: Ana

# Podemos alterar um valor existente
pessoa["idade"] = 31

# Podemos adicionar uma nova chave
pessoa["email"] = "ana@email.com"

print(pessoa)
# Saída: {'nome': 'Ana', 'idade': 31, 'cidade': 'Recife', 'email': 'ana@email.com'}
As chaves precisam ser hasháveis — normalmente textos (str), números ou tuplas. Listas e outros dicionários não podem ser chaves, porque são mutáveis.
```

### 2 Criar dicionários

```python
# 1) Forma mais comum: chaves e valores literais
produto = {"nome": "Teclado", "preco": 199.90}

# 2) Dicionário vazio, útil para preencher depois
carrinho = {}

# 3) A partir de pares (chave, valor)
cores = dict([("primaria", "azul"), ("secundaria", "amarelo")])

# 4) Usando argumentos nomeados
config = dict(debug=True, porta=8000)

# 5) Dict comprehension: cria um dicionário com uma regra
quadrados = {numero: numero ** 2 for numero in range(1, 6)}
# Resultado: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
O construtor dict() aceita pares iteráveis e argumentos nomeados; se uma chave se repetir, o último valor atribuído prevalece.
```

### 3 Ler valores com segurança

```python
usuario = {"nome": "Bruno", "idioma": "pt-BR"}

# Acesso direto: gera KeyError se a chave não existir
print(usuario["nome"])  # Bruno

# Método get(): retorna None (ou um valor padrão) se a chave não existir
print(usuario.get("telefone"))               # None
print(usuario.get("telefone", "Não informado"))  # Não informado

# Verificar se uma chave existe antes de acessar
if "idioma" in usuario:
    print("Idioma configurado:", usuario["idioma"])
Prefira get() quando a ausência da chave for uma situação normal; use dicionario["chave"] quando a chave for obrigatória e você quiser que o erro apareça claramente. O operador in verifica chaves de forma direta.
```

### 4 Atualizar e remover

```python
estoque = {"arroz": 10, "feijao": 5}

# Atualiza um valor existente
estoque["arroz"] = 8

# Adiciona um novo item
estoque["macarrao"] = 12

# Remove um item e devolve o valor removido
quantidade = estoque.pop("feijao")
print(quantidade)  # 5

# Remove o último item inserido e o retorna
ultimo = estoque.popitem()

# Remove todos os itens
estoque.clear()

# Remove uma chave com del; gera KeyError se não existir
produtos = {"cafe": 3}
del produtos["cafe"]
del d[chave], pop() e popitem() alteram o dicionário; pop() permite informar um valor padrão para evitar erro quando a chave não existe.

Percorrer dicionários
python
notas = {"Ana": 9.5, "Bruno": 7.0, "Carla": 8.8}

# Percorre apenas as chaves
for aluno in notas:
    print(aluno, notas[aluno])

# Forma recomendada: percorre chave e valor juntos
for aluno, nota in notas.items():
    print(f"{aluno} tirou {nota}")

# Obtém apenas valores
for nota in notas.values():
    print(nota)

# Obtém apenas chaves
for aluno in notas.keys():
    print(aluno)
Os métodos keys(), values() e items() retornam visões dinâmicas: se o dicionário mudar, a visão reflete a mudança.
```

### 5 Juntar dicionários

```python
padrao = {"tema": "claro", "idioma": "pt-BR"}
preferencias = {"tema": "escuro", "notificacoes": True}

# O operador | cria um novo dicionário
configuracao = padrao | preferencias

# Se houver chaves repetidas, o dicionário da direita prevalece
print(configuracao)
# {'tema': 'escuro', 'idioma': 'pt-BR', 'notificacoes': True}

# update() altera o dicionário original
padrao.update(preferencias)
O operador | combina dois dicionários em um novo objeto, e update() aplica as alterações no dicionário da esquerda. 
```

### 6 Erros comuns

```python
# ERRADO: usar uma lista como chave
# dados = {["a", "b"]: 1}  # TypeError: unhashable type: 'list'

# CORRETO: usar uma tupla, que é imutável
dados = {("a", "b"): 1}

# ERRADO: acessar chave inexistente sem proteção
config = {}
# print(config["porta"])  # KeyError: 'porta'

# CORRETO: usar get() com valor padrão
print(config.get("porta", 8000))  # 8000

# Cuidado: chaves repetidas não criam dois itens
repetido = {"status": "ativo", "status": "inativo"}
print(repetido)  # {'status': 'inativo'}
Valores que se comparam como iguais — por exemplo, 1, 1.0 e True — podem se referir à mesma entrada quando usados como chave.
```

### 7 Dicionários aninhados

```python
# Um dicionário pode conter listas e outros dicionários
escola = {
    "nome": "Escola Python",
    "turmas": {
        "python-basico": {
            "professor": "Marcos",
            "alunos": 25,
        },
        "django-avancado": {
            "professor": "Julia",
            "alunos": 18,
        },
    },
}

# Acessando dados em níveis mais profundos
print(escola["turmas"]["django-avancado"]["professor"])  # Julia

# Alterando um valor interno
escola["turmas"]["python-basico"]["alunos"] = 27
Essa estrutura é muito útil para representar objetos do mundo real, respostas de APIs e configurações.
```

### 8 Dicionários em Django 6

No Django, dicionários aparecem com frequência em contextos de template, configurações e dados enviados pelo usuário. O Django 6.0 suporta Python 3.12, 3.13 e 3.14, e usa dict e QueryDict como mapeamentos em recursos como a tag querystring.

Enviando dados para um template

```python
# views.py
from django.shortcuts import render

def painel(request):
    # O contexto é um dicionário: a chave vira uma variável no template
    contexto = {
        "titulo": "Painel do aluno",
        "aluno": {
            "nome": "Ana",
            "media": 9.2,
        },
        "disciplinas": ["Python", "Django", "Banco de dados"],
    }

    # O Django disponibiliza cada chave do contexto no template
    return render(request, "painel.html", contexto)
xml
<!-- templates/painel.html -->
<h1>{{ titulo }}</h1>

<p>Aluna: {{ aluno.nome }}</p>
<p>Média: {{ aluno.media }}</p>

<ul>
  {% for disciplina in disciplinas %}
    <li>{{ disciplina }}</li>
  {% endfor %}
</ul>
Aqui, aluno é um dicionário Python e aluno.nome acessa a chave "nome" dentro do template.

Lendo dados de formulários
python
# views.py
from django.http import HttpResponse
from django.shortcuts import render

def cadastro(request):
    if request.method == "POST":
        # request.POST é um QueryDict: funciona de forma parecida com um dicionário
        nome = request.POST.get("nome", "").strip()
        email = request.POST.get("email", "").strip()

        if not nome or not email:
            return render(
                request,
                "cadastro.html",
                {"erro": "Informe nome e e-mail."},
            )

        return HttpResponse(f"Cadastro recebido para {nome}.")

    return render(request, "cadastro.html")
xml
<!-- templates/cadastro.html -->
<form method="post">
  {% csrf_token %}
  <input type="text" name="nome" placeholder="Nome">
  <input type="email" name="email" placeholder="E-mail">
  <button type="submit">Enviar</button>
</form>

{% if erro %}
  <p>{{ erro }}</p>
{% endif %}
Use request.POST.get("campo") para evitar KeyError quando o campo não foi enviado. Em Django, QueryDict é um tipo de mapeamento semelhante a dict, mas preparado para dados de formulários e query strings.
```

### 9 Boas práticas
Use nomes de chaves claros e consistentes, como "nome_completo" em vez de "n".

Prefira get() para campos opcionais e acesso direto para campos obrigatórios.

Não use listas ou dicionários como chaves; prefira str, int, float, bool ou tuple.

Para estruturas complexas e fixas, considere classes, dataclasses ou modelos Django, que deixam a intenção mais explícita.

Em Django, valide dados recebidos do usuário com formulários ou serializers antes de usá-los.

Exercício prático

```python
# Crie um mini sistema de contato usando um dicionário
contatos = {}

def adicionar_contato(contatos, nome, telefone):
    # A chave é o nome; o valor é o telefone
    contatos[nome] = telefone

def buscar_contato(contatos, nome):
    # Retorna uma mensagem amigável caso o contato não exista
    return contatos.get(nome, "Contato não encontrado.")

adicionar_contato(contatos, "Ana", "99999-0000")
adicionar_contato(contatos, "Bruno", "98888-1111")

print(buscar_contato(contatos, "Ana"))     # 99999-0000
print(buscar_contato(contatos, "Carla"))   # Contato não encontrado.
Esse exemplo mostra o principal uso de um dicionário: relacionar um identificador único a uma informação e recuperá-la rapidamente.
```
