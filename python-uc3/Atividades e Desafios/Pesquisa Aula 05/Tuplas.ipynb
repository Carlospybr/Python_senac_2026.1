```python
Guia de Tuplas em Python 3
Uma tupla é uma coleção ordenada de valores que, depois de criada, não pode ser alterada. Ela é semelhante a uma lista, mas é imutável — ideal para representar dados fixos, como coordenadas, datas ou pares chave-valor.

O que é uma tupla?
Pense em uma tupla como uma caixa lacrada: você pode olhar o que está dentro, mas não pode trocar, acrescentar ou remover itens. Em Python, tuplas são sequências imutáveis e podem conter valores de tipos diferentes.

python
# Criação básica de uma tupla
coordenada = (10, 20)

print(coordenada)          # Saída: (10, 20)
print(type(coordenada))    # Saída: <class 'tuple'>
Criando tuplas
Os parênteses ajudam na leitura, mas o que realmente define uma tupla é a vírgula. Para uma tupla de um único elemento, a vírgula é obrigatória.

python
# Tupla com vários elementos
pessoa = ("Ana", 30, "São Paulo")

# Tupla sem parênteses: a vírgula cria a tupla
cores = "azul", "verde", "vermelho"

# Tupla com um único elemento
um_item = (42,)        # Correto: observe a vírgula
nao_e_tupla = (42)     # Errado: isto é apenas o número 42

# Tupla vazia
vazia = ()

# Conversão de lista para tupla
numeros = tuple([1, 2, 3])

print(pessoa)       # ('Ana', 30, 'São Paulo')
print(cores)        # ('azul', 'verde', 'vermelho')
print(um_item)      # (42,)
print(numeros)      # (1, 2, 3)
Acessando elementos
Cada item tem uma posição chamada índice. A contagem começa em 0; índices negativos contam a partir do final.

python
frutas = ("maçã", "banana", "uva", "laranja")

print(frutas[0])    # 'maçã' — primeiro elemento
print(frutas[2])    # 'uva' — terceiro elemento
print(frutas[-1])   # 'laranja' — último elemento

# Fatiamento: pega do índice 1 até o 3, sem incluir o 3
print(frutas[1:3])  # ('banana', 'uva')

# len() informa quantos elementos existem
print(len(frutas))  # 4
Imutabilidade
Tentar alterar um item de uma tupla gera um erro TypeError, porque tuplas não suportam atribuição por índice.

python
ponto = (10, 20)

# Isto gera erro:
# ponto[0] = 99
# TypeError: 'tuple' object does not support item assignment

# Para "alterar", crie uma nova tupla
ponto_atualizado = (99, ponto[1])

print(ponto_atualizado)  # (99, 20)
Atenção: a tupla em si é imutável, mas ela pode conter objetos mutáveis, como listas. Nesse caso, o conteúdo interno da lista pode mudar.

python
dados = ("nome", ["python", "django"])

# A tupla continua igual, mas a lista interna pode ser modificada
dados[1].append("postgresql")

print(dados)  # ('nome', ['python', 'django', 'postgresql'])
Desempacotamento
Desempacotar é distribuir os valores de uma tupla em variáveis. É um dos usos mais práticos e legíveis de tuplas.

python
# Empacotamento: valores viram uma tupla
registro = ("Maria", 28, "Engenheira")

# Desempacotamento: cada valor vai para uma variável
nome, idade, profissao = registro

print(nome)        # Maria
print(idade)       # 28
print(profissao)   # Engenheira
Você pode usar * para agrupar os elementos restantes:

python
numeros = (1, 2, 3, 4, 5)

# O primeiro vai para 'primeiro'; o resto vai para 'resto'
primeiro, *resto = numeros

print(primeiro)  # 1
print(resto)     # [2, 3, 4, 5]
Troca de valores
Tuplas permitem trocar valores entre variáveis sem criar uma variável temporária:

python
a = 1
b = 2

# Python cria temporariamente a tupla (b, a) e a desempacota
a, b = b, a

print(a)  # 2
print(b)  # 1
Métodos úteis
Tuplas têm poucos métodos justamente porque são imutáveis. Os principais são count() e index().

python
letras = ("a", "b", "a", "c", "a")

# count() conta quantas vezes um valor aparece
print(letras.count("a"))  # 3

# index() retorna a posição da primeira ocorrência
print(letras.index("b"))  # 1
Tuplas aninhadas
Uma tupla pode conter outras tuplas, formando estruturas mais complexas:

python
# Cada tupla interna representa uma coordenada (x, y)
pontos = ((0, 0), (3, 4), (6, 8))

print(pontos[1])       # (3, 4)
print(pontos[1][0])    # 3 — primeiro valor da segunda tupla
print(pontos[1][1])    # 4 — segundo valor da segunda tupla
Tuplas como chaves de dicionário
Tuplas podem ser usadas como chaves em dicionários quando contêm apenas valores imutáveis, como números, textos ou outras tuplas. Listas não podem ser chaves porque podem ser modificadas.

python
# Tupla (latitude, longitude) como chave
locais = {
    (-23.5505, -46.6333): "São Paulo",
    (-22.9068, -43.1729): "Rio de Janeiro",
}

print(locais[(-23.5505, -46.6333)])  # São Paulo
Tupla ou lista?
Use tupla quando os dados representam uma estrutura fixa; use lista quando precisar adicionar, remover ou alterar itens. A documentação oficial destaca que tuplas são imutáveis e frequentemente heterogêneas, enquanto listas são mutáveis e normalmente homogêneas.

Situação	Recomendação	Exemplo
Coordenadas fixas	Tupla	(latitude, longitude)
Lista de tarefas que muda	Lista	["estudar", "programar"]
Retorno com vários valores	Tupla	return nome, idade
Coleção que cresce em tempo de execução	Lista	usuarios.append(novo)
Chave de dicionário composta	Tupla	{(2026, 10): "outubro"}
Tuplas em funções
Funções podem retornar vários valores de forma natural usando tuplas:

python
def dividir_e_resto(dividendo, divisor):
    # Retorna dois valores: quociente e resto
    quociente = dividendo // divisor
    resto = dividendo % divisor
    return quociente, resto  # Isto é uma tupla


resultado = dividir_e_resto(17, 5)

print(resultado)  # (3, 2)

# Também é possível desempacotar direto
quociente, resto = dividir_e_resto(17, 5)

print(quociente)  # 3
print(resto)      # 2
Tuplas nomeadas
Para código mais legível, namedtuple cria tuplas em que cada posição tem um nome. É útil quando a tupla representa um registro, como uma pessoa ou um produto.

python
from collections import namedtuple

# Define um tipo de tupla com campos nomeados
Pessoa = namedtuple("Pessoa", ["nome", "idade", "cidade"])

# Cria uma instância
ana = Pessoa(nome="Ana", idade=30, cidade="Recife")

# Acesso por índice continua funcionando
print(ana[0])       # Ana

# Mas o acesso pelo nome é mais claro
print(ana.nome)     # Ana
print(ana.idade)    # 30
print(ana.cidade)   # Recife
Tuplas e Django 6
Em Django 6.0, tuplas aparecem com frequência em configurações, modelos e consultas. O Django 6.0 suporta Python 3.12, 3.13 e 3.14.

python
# settings.py — escolhas fixas para um campo de modelo
STATUS_CHOICES = (
    ("pendente", "Pendente"),
    ("pago", "Pago"),
    ("cancelado", "Cancelado"),
)
python
# models.py — usando as escolhas em um campo
from django.db import models


class Pedido(models.Model):
    # Cada tupla representa (valor_salvo_no_banco, rótulo_exibido)
    status = models.CharField(
        max_length=20,
        choices=STATUS_CHOICES,
        default="pendente",
    )
python
# views.py — uma view pode retornar vários valores em uma tupla
from django.http import JsonResponse


def resumo(request):
    total_pedidos = 120
    total_clientes = 45

    # Retorna uma tupla com dois valores
    return JsonResponse(
        {
            "pedidos": total_pedidos,
            "clientes": total_clientes,
        }
    )
Erros comuns
python
# ERRO: tupla de um item sem vírgula
valor = (10)
print(type(valor))  # <class 'int'>, não tuple

# CORRETO
valor = (10,)
print(type(valor))  # <class 'tuple'>
python
# ERRO: tentar modificar uma tupla
cores = ("azul", "verde")

# cores.append("amarelo")
# AttributeError: 'tuple' object has no attribute 'append'

# CORRETO: criar uma nova tupla
cores = cores + ("amarelo",)
print(cores)  # ('azul', 'verde', 'amarelo')
Regras práticas
Use tuplas para dados que devem permanecer fixos.

Use listas quando a coleção precisar mudar.

Lembre-se: a vírgula cria a tupla; os parênteses são, em muitos casos, apenas organização visual.

Prefira desempacotamento para deixar o código mais legível.

Em Django, use tuplas para conjuntos fixos de opções, como choices.

Não use tuplas contendo listas como chaves de dicionário, pois parte do conteúdo seria mutável.
```
