# Guia de Listas em Python 3

### 1. Uma lista é uma coleção ordenada e mutável de valores. Ela é uma das estruturas mais usadas em Python: serve para guardar nomes, produtos, números, resultados de consultas e muitos outros conjuntos de dados.

Importante: este guia usa Python 3. O Django 6.0 oficialmente suporta Python 3.12, 3.13 e 3.14, portanto os exemplos funcionam nesse contexto.

O que é uma lista
Uma lista é escrita entre colchetes, com os itens separados por vírgula. Ela mantém a ordem de inserção e pode ser alterada depois de criada — por isso é chamada de mutável.

```python
# Criação de uma lista simples com nomes de frutas
frutas = ["maçã", "banana", "laranja"]

# Uma lista pode conter tipos diferentes,
# embora seja mais comum manter itens do mesmo tipo
dados_mistos = ["Ana", 25, True, 3.14]

# Lista vazia: útil quando os itens serão adicionados depois
tarefas = []

# Exibindo as listas
print(frutas)
print(dados_mistos)
print(tarefas)
```

### 2 Acessando itens
#### Cada item tem uma posição chamada índice. Em Python, a contagem começa em 0, não em 1.

```python

frutas = ["maçã", "banana", "laranja"]

# O índice 0 representa o primeiro item
print(frutas[0])   # Saída: maçã

# O índice 1 representa o segundo item
print(frutas[1])   # Saída: banana

# O índice -1 representa o último item
print(frutas[-1])  # Saída: laranja

# Acessar um índice inexistente gera IndexError
# print(frutas[10])  # Erro: IndexError
```

### 3 Fatiamento
O fatiamento permite pegar uma parte da lista. O formato é lista[início:fim]; o item na posição fim não entra no resultado.


```python
numeros = [10, 20, 30, 40, 50, 60]

# Pega do índice 1 até o índice 3 (o 4 não entra)
print(numeros[1:4])   # Saída: [20, 30, 40]

# Pega os três primeiros itens
print(numeros[:3])    # Saída: [10, 20, 30]

# Pega do terceiro item até o final
print(numeros[2:])    # Saída: [30, 40, 50, 60]

# Pega os três últimos itens
print(numeros[-3:])   # Saída: [40, 50, 60]
```

### 4 Alterando itens
Como listas são mutáveis, você pode substituir qualquer item pelo índice.


```python
cores = ["azul", "verde", "vermelho"]

# Substitui o item na posição 1
cores[1] = "amarelo"

print(cores)  # Saída: ['azul', 'amarelo', 'vermelho']
Adicionando itens
Use append() para acrescentar um item ao final e extend() para acrescentar vários itens de outra coleção.

animais = ["cachorro", "gato"]

# append() adiciona um único item ao final
animais.append("coelho")
print(animais)  # Saída: ['cachorro', 'gato', 'coelho']

# extend() adiciona cada item de outra lista separadamente
animais.extend(["papagaio", "peixe"])
print(animais)
# Saída: ['cachorro', 'gato', 'coelho', 'papagaio', 'peixe']

# insert() adiciona um item em uma posição específica
animais.insert(0, "hamster")
print(animais)
# Saída: ['hamster', 'cachorro', 'gato', 'coelho', 'papagaio', 'peixe']
Cuidado: append(["a", "b"]) adiciona a lista inteira como um único item; extend(["a", "b"]) adiciona "a" e "b" separadamente.


lista = [1, 2]

# Adiciona uma lista como um único elemento
lista.append([3, 4])
print(lista)  # Saída: [1, 2, [3, 4]]

# Adiciona os elementos individualmente
outra = [1, 2]
outra.extend([3, 4])
print(outra)  # Saída: [1, 2, 3, 4]
Removendo itens
Python oferece formas diferentes de remover dados, conforme você conhece o valor ou o índice.


frutas = ["maçã", "banana", "laranja", "banana"]

# remove() exclui a primeira ocorrência do valor informado
frutas.remove("banana")
print(frutas)  # Saída: ['maçã', 'laranja', 'banana']

# pop() remove e retorna um item; sem índice, remove o último
ultima_fruta = frutas.pop()
print(ultima_fruta)  # Saída: banana
print(frutas)        # Saída: ['maçã', 'laranja']

# del remove pelo índice, mas não devolve o valor removido
del frutas[0]
print(frutas)  # Saída: ['laranja']

# clear() esvazia a lista
frutas.clear()
print(frutas)  # Saída: []
Medindo e procurando
Use len() para saber quantos itens existem, in para verificar existência e count() para contar repetições.


numeros = [5, 3, 8, 3, 9, 3]

# len() retorna a quantidade de itens
print(len(numeros))  # Saída: 6

# in verifica se um valor existe na lista
print(8 in numeros)   # Saída: True
print(7 in numeros)   # Saída: False

# count() conta quantas vezes um valor aparece
print(numeros.count(3))  # Saída: 3

# index() retorna o índice da primeira ocorrência
print(numeros.index(8))  # Saída: 2
Ordenando listas
sort() ordena a própria lista, enquanto sorted() cria uma nova lista ordenada, preservando a original.


notas = [7.5, 9.0, 6.0, 8.5]

# sort() altera a lista original
notas.sort()
print(notas)  # Saída: [6.0, 7.5, 8.5, 9.0]

# reverse=True ordena do maior para o menor
notas.sort(reverse=True)
print(notas)  # Saída: [9.0, 8.5, 7.5, 6.0]

# sorted() não altera a lista original
originais = [4, 1, 3]
ordenadas = sorted(originais)

print(originais)  # Saída: [4, 1, 3]
print(ordenadas)  # Saída: [1, 3, 4]
Percorrendo listas
O for é a forma mais comum de visitar cada item de uma lista.


nomes = ["Ana", "Bruno", "Carla"]

# Percorre cada nome da lista
for nome in nomes:
    print(f"Olá, {nome}!")

# Saída:
# Olá, Ana!
# Olá, Bruno!
# Olá, Carla!
Se você também precisar do índice, use enumerate().


nomes = ["Ana", "Bruno", "Carla"]

# enumerate() entrega o índice e o valor ao mesmo tempo
for indice, nome in enumerate(nomes, start=1):
    print(f"{indice}. {nome}")

# Saída:
# 1. Ana
# 2. Bruno
# 3. Carla
```

### 5 Compreensão de listas
A compreensão de lista cria uma lista nova de forma concisa, aplicando uma expressão a cada item e, opcionalmente, filtrando com if.


```python
numeros = [1, 2, 3, 4, 5, 6]

# Forma tradicional com for
pares_tradicional = []
for numero in numeros:
    if numero % 2 == 0:
        pares_tradicional.append(numero)

# Mesma ideia usando compreensão de lista
pares = [numero for numero in numeros if numero % 2 == 0]

print(pares)  # Saída: [2, 4, 6]

# Aplica uma operação a cada item
dobros = [numero * 2 for numero in numeros]
print(dobros)  # Saída: [2, 4, 6, 8, 10, 12]
```

### 6 Cópias: atenção à referência
Atribuir uma lista a outra variável não cria uma cópia independente. As duas variáveis passam a apontar para a mesma lista na memória.


```python
originais = [1, 2, 3]

# Isso NÃO cria uma cópia independente
referencia = originais
referencia.append(4)

print(originais)    # Saída: [1, 2, 3, 4]
print(referencia)   # Saída: [1, 2, 3, 4]

# Para criar uma cópia rasa, use copy() ou fatiamento
copia = originais.copy()
copia.append(5)

print(originais)  # Saída: [1, 2, 3, 4]
print(copia)      # Saída: [1, 2, 3, 4, 5]
```

### 7 Uma cópia rasa copia a lista externa, mas não duplica listas internas. Para estruturas profundamente aninhadas, avalie copy.deepcopy().

Listas aninhadas
Uma lista pode conter outras listas, formando algo parecido com uma tabela.


```python
# Cada lista interna representa uma linha de uma tabela
alunos = [
    ["Ana", 8.5],
    ["Bruno", 7.0],
    ["Carla", 9.2],
]

# Acessa a primeira linha
print(alunos[0])       # Saída: ['Ana', 8.5]

# Acessa o nome da primeira aluna
print(alunos[0][0])    # Saída: Ana

# Percorre todas as linhas
for nome, nota in alunos:
    print(f"{nome} tirou {nota}")
Erros comuns
Erro	Exemplo	Como evitar
IndexError	frutas[10] em uma lista pequena	Verifique len(lista) antes de acessar
ValueError	lista.remove("x") quando "x" não existe	Confirme com if "x" in lista:
Alterar a lista durante o for	Remover itens enquanto percorre	Percorra uma cópia: for item in lista.copy():
Confundir append() com extend()	lista.append([1, 2])	Use extend() para juntar itens individualmente
Achar que = copia a lista	copia = original	Use original.copy() ou original[:]
python
# Remoção segura durante a leitura
numeros = [1, 2, 3, 4, 5, 6]

# Percorre uma cópia para poder remover da lista original
for numero in numeros.copy():
    if numero % 2 == 0:
        numeros.remove(numero)

print(numeros)  # Saída: [1, 3, 5]
```

### 8 Listas no contexto Django
Em Django, listas aparecem com frequência ao manipular dados antes de salvá-los, preparar opções para formulários ou organizar resultados. Porém, para conjuntos grandes vindos do banco de dados, prefira trabalhar com QuerySets e avaliar a lista somente quando necessário.


```python
# Exemplo didático em uma view Django
from django.shortcuts import render

def lista_de_tarefas(request):
    # Em uma aplicação real, estes dados normalmente viriam de um model:
    # Tarefa.objects.all()
    tarefas = [
        {"titulo": "Estudar Python", "concluida": True},
        {"titulo": "Criar um app Django", "concluida": False},
        {"titulo": "Escrever testes", "concluida": False},
    ]

    # Filtra as tarefas pendentes usando compreensão de lista
    pendentes = [
        tarefa for tarefa in tarefas
        if not tarefa["concluida"]
    ]

    # Envia a lista para o template
    return render(
        request,
        "tarefas/lista.html",
        {"tarefas": tarefas, "pendentes": pendentes},
    )
xml
<!-- templates/tarefas/lista.html -->
<h1>Minhas tarefas</h1>

<ul>
  {% for tarefa in tarefas %}
    <li>
      {{ tarefa.titulo }}
      {% if tarefa.concluida %}
        — concluída
      {% else %}
        — pendente
      {% endif %}
    </li>
  {% empty %}
    <li>Nenhuma tarefa cadastrada.</li>
  {% endfor %}
</ul>

<p>Total de pendentes: {{ pendentes|length }}</p>
Boas práticas
Prefira listas com itens do mesmo tipo quando isso deixar o código mais previsível.

Use in antes de remove() ou index() quando o valor pode não existir.

Use sorted() quando quiser preservar a ordem original.

Use copy() quando precisar de uma lista independente.

Use compreensão de lista para transformações simples; para lógica complexa, um for tradicional costuma ficar mais legível.

Em Django, não transforme QuerySets em listas sem necessidade: isso pode carregar mais dados do que o necessário.
```
