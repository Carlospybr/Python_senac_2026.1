### 3️⃣ Imprimir Números pares e impares
### 🔖 Elabore um código para imprimir todos os números pares e ímpares de 1 a 20 usando for

```python
pares=[]
impares=[]

for numero in range (1,21):
    if (numero % 2 == 0 ):
        pares.append(numero)
    else:
        impares.append(numero)

print(f'Os números pares são: {pares}')
print(f'Os números impares são: {impares}')
```

### 4️⃣ Imprimir de 10 até 1
### 🔖 Elabore um código para imprimir de 10 até 1 de forma decrescente, usando while

#### Fatorial usando While

```python
numero = 10
neutro = 1  # Iniciamos com 1, o elemento neutro da multiplicação

# O loop roda enquanto o número for maior que 1
while numero > 1:
    # Multiplicamos o valor acumulado no 'neutro' pelo 'numero' da rodada atual
    neutro = neutro * numero

    # Reduzimos o valor do número em 1 para a próxima rodada
    numero = numero - 1

print(neutro)
```

#### Contagem de dígitos (while)


```python
# Versão Pythonica

while True:
    try:
        numero = int(input('Insira um numero inteiro: '))
        break
    except ValueError:
        print('Valor invalido! Tente novamente.')

contagem = len(str(abs(numero)))

print(contagem)
```

```python
# Versão Algorítmica (Raciocínio Lógico)

def contar_digitos(numero):
    numero = abs(int(numero))
    contador = 0

    if numero == 0:
        return 1

    while numero > 0:
        numero = numero // 10
        contador = contador + 1

    return contador

while True:
        try:
            numero = int(input('Insira um numero: '))
            break
        except ValueError:
            print('Valor invalido! Tente novamente.')
```
