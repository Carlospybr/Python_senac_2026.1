# 7

```python
km = float(input("Insira o Km da viagem : "))
```

```python
vm = float(input("Insira a sua velocidade média programada para a viagem : "))
```

```python
t = km / vm
```

```python
if  t <=   0.59   : 
    print(f'O tempo para chegar ao seu destino será de, aproximadamente : {t:.2f} minutos.')
elif     t ==   0.60   : 
    print(f'O tempo para chegar ao seu destino será de, aproximadamente : {t:.0f} hora.')
elif    t >=   0.61:  
    print(f'O tempo para chegar ao seu destino será de, aproximadamente : {t:.0f} horas e {t:0.1f} minuto')
else :
    print(f'O tempo para chegar ao seu destino será de, aproximadamente : {t:.0f} horas e {t:0.2f} minutos')
```
