```python
kg = float(input("Insira o seu peso :"))
alt = float(input("Insira a sua altura :"))
```

```python
imc = kg / alt**2
```

```python
if imc < 18.5 : 
    res = "abaixo do peso"
elif imc < 24.9:
    res = "no peso ideal."
elif imc < 29.9   :
    res = "sobrepeso"
else :
    res = "obeso"
```

```python
print(f'Seu IMC é {imc:.2f} Portanto, você está {res}')
```
