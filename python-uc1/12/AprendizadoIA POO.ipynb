```python
class ContaBancaria:
    # Método Construtor (inicializa os atributos da instância)
    def __init__(self, titular, saldo_inicial=0.0):
        self.titular = titular
        self.saldo = saldo_inicial

    # Método (comportamento)
    def depositar(self, valor):
        if valor > 0:
            self.saldo += valor
            return f"Depósito de R$ {valor:.2f} realizado com sucesso."
        return "Valor inválido para depósito."

# Criando uma instância (objeto)
minha_conta = ContaBancaria("Ana", 1000.0)
print(minha_conta.depositar(500.0))
print(f"Saldo atual de {minha_conta.titular}: R$ {minha_conta.saldo:.2f}")
```
