```python
chaves = {"nome", "dt_nascimento", "cep", "logradouro", "bairro", "cidade", "estado"}
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)
print(usuario)
```

```python
usuario["nome"]="Carlos"
```

```python
usuario["nome"]
usuario["dt_nascimento"]="25/09/2003"
usuario["cep"]="257481955"
usuario["logradouro"]="Rua das Flores"
usuario["bairro"]="Cuiaba"
usuario["cidade"]="Petropolis"
usuario["estado"]="RJ"
```

```python
for chaves, valor in usuario.items():
    print(f'{chaves.capitalize():_<20}: {valor.capitalize()}')
```
