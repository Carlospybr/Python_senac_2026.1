```python
def saudação(display="sem nome"):
    print (f'Olá, {display}!')
```

```python
saudação('Carlos')
```

```python
saudação()
```

## Passagem de argumentos arbritários

```python
def saudações(*nomes): 
    #print(nomes)
    for nome in nomes :
        saudação(nome)
```

```python
#pessoas=()
saudações('Carlos', 'Rodrigo', 'Guilherme')
```
