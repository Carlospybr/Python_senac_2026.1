# Par ou Ímpar

# Fatorial

```python
def fatorial(n):
    if n < 0:
        print('Número inválido.')
    elif n == 0:
        return 1
    else:
        resultado = 1
        for i in range(1, n + 1):
            resultado *= i 
        return resultado

print('Fatorial de 5:', fatorial(5))
```
