#### Criando um vetor (lista) com números inteiros

```python
vetor = [10, 20, 30, 40, 50]
print(vetor)
```

#### Iterando sobre uma lista

```python
for elemento in vetor:
    print("Elemento:", elemento)
```

#### Utilizando NumPy para operações com vetores

```python
import numpy as np
vetor_np = np.array([1, 2, 3, 4, 5])
print("Vetor com NumPy:", vetor_np)
```

#### Calcular a soma dos elementos de um vetor


```python
vetor = [10, 20, 30, 40, 50]
soma =sum(vetor)
print(soma)
```

#### Encontrar o maior e o menor valor em um vetor

```python
vetor = [10, 20, 30, 40, 50]

maior = max(vetor)
menor = min(vetor)

print(maior, menor)
```

#### Inverter a ordem dos elementos de um vetor sem usar reverse()

```python
vetor = [10, 20, 30, 40, 50]

print(vetor[::-1])
```

#### Multiplicar cada elemento do vetor por 2 e armazenar em um novo vetor

```python
vetor = [10, 20, 30, 40, 50]
vetor_mult = []
for i in vetor:
    vetor_mult.append(i**2)

print(vetor_mult)
```

#### Contar quantas vezes o valor 3 aparece em um vetor

```python
vetor = [3, 1, 4, 3, 5, 3]
n = []
for i in vetor:
    n = i
    if n == i:
        n = n+1
print(n)
```
