#### Cadastro de produtos

```python
produtos = (
    ("Arroz", 5.99, 10 ),
    ("Feijão", 7.49, 7),
    ("Leite", 4.89, 2),
    ("Óleo", 9.99, 5),
    ("Açúcar", 3.29, 8)
)
valor_produto=0
valor_total=0

print("LISTA DE PRODUTOS")

for nome, preco, qt in produtos:
    valor_produto= qt*preco
    valor_total+=valor_produto
    print(f"{nome:.<20}, R$ {preco:5.2f} .. QT {qt:3d} .. Total: R$ {valor_produto:2.2f}")

print(f"\n\t valor total: R$       {valor_total}")
```

#### Números Primos

```python
numeros = (2, 7, 10, 13, 17, 22, 29, 31, 40, 53)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primos = tuple(num for num in numeros if eh_primo(num))
print(f"Números primos: {primos}")
```
