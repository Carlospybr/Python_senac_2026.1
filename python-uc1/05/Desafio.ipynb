# Jogo de Adivinhação

```python
import random
segredo=random.randint(1,10)
for _ in range(3):
    numero=int(input('Digite um numero: '))
    if numero==segredo :
        print('Parabéns! Você acertou!')
        break
    else:
        print('Tente Novamente!')
print(f'O número secreto era {segredo}')
```
